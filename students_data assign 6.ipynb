```python
import pandas as pd


sheet_url = "https://docs.google.com/spreadsheets/d/19IqlRhmifKFIOi7CnM0eoRpXHZg1PcWvFfkgSIGVdT0/export?format=xlsx"

# Read the Excel file directly into a Pandas DataFrame
df = pd.read_excel(sheet_url)


df.to_excel("Students_data.xlsx", index=False)

print("--- Original Dataset ---")
print(df)

# --- Data Exploration ---
print("\n--- First 5 Rows (head) ---")
print(df.head())

print("\n--- Dataset Info (info) ---")
print(df.info())

print("\n--- Statistical Summary (describe) ---")
print(df.describe())

# --- Handle Missing Values ---
print("\n--- Missing Values Count ---")
print(df.isnull().sum())

# Fill missing marks with mean, missing city with mode
df['marks'] = df['marks'].fillna(df['marks'].mean())
df['city'] = df['city'].fillna(df['city'].mode()[0])

# Drop rows where 'name' is missing
df = df.dropna(subset=['name'])

# --- Remove Duplicates ---
print("\n--- Duplicate Rows Found ---")
print(df.duplicated().sum())
df = df.drop_duplicates()

# --- Data Selection & Filtering using loc[] ---
high_scorers = df.loc[df['marks'] > 70, ['roll no', 'name', 'marks', 'city']]

print("\n--- Final Cleaned & Filtered Data (marks > 70) ---")
print(high_scorers)
```
