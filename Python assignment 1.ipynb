```python
# 7 Calculate frequency of letters occouring in a given string
e='Mississipi' 
freq={}
for characters in e:
    freq[characters]=freq.get(characters,0)+1

    print('Frequency of letters:',freq)
```

```python
# 9no of days between the 2 date 
 from datetime import date
d11=(1,1,2002)
d12=(29,8,2026)
date1=date(d11[0],d11[1],d11[2])
date2=date(d12[0],d12[1],d12[2])

daysbetween=abs((date2-date1).days)
print('days between dates:',daysbetween)
```
